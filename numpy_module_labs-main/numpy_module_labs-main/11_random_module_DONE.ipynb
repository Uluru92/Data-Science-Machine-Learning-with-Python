
6. Generate a 1D array with 6 random floats from a standard normal distribution (mean=0, std=1) using `np.random.standard_normal()`.


7. Generate a 2D array with shape (4, 3) containing random floats from a normal distribution with mean=5 and std=2 using `np.random.normal()`.


```python
a = np.random.normal(loc=5,scale=2,size=(4,3))
print(f'2D array:\n{a}')

import matplotlib.pyplot as plt
from scipy.stats import norm

a = np.random.normal(loc=5, scale=2, size=(4000,))  # use more data for a smoother curve

# Histogram (density=True normalizes it)
count, bins, ignored = plt.hist(a, bins=30, density=True, color='lightblue', edgecolor='black')

# Gaussian curve
x = np.linspace(min(a), max(a), 100)
plt.plot(x, norm.pdf(x, np.mean(a), np.std(a)), 'r-', linewidth=2, label='Gaussian Curve')

plt.title("Gaussian Bell Curve (Normal Distribution)")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```


8. Create a list of 5 colors and use `np.random.choice()` to randomly select 3 colors from the list with equal probability.


```python
list_colors = ['red','green','yellow','purple','black']
random_selection = np.random.choice(list_colors)
print(f'Random selection from list of colors: color {random_selection}')
```


9. From the same list of colors, randomly select 4 colors with replacement and equal probability using `np.random.choice()`.


```python
random_selection = np.random.choice(list_colors, size=4, replace=True)
print(f'4 random colors selected: {random_selection}')
```


10. From the same list of colors, randomly select 2 colors without replacement and with specified probabilities using `np.random.choice()`.


```python
random_selection = np.random.choice(list_colors, size=2, replace=False, p=[0.9,0.025,0.025,0.025,0.025])
print(f'2 random colors selected: {random_selection}')
```
